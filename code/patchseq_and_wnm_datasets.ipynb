# Reference Tasic 2018 data set

```python
tasic_2018_ds = DataSet(
    id="tasic_2018_visp_scrnaseq",
    name="Tasic et al. 2018 VISp RNA-seq data set",
    publication="doi.org/10.1038/s41586-018-0654-5",
    modality=Modality.OTHER.value,
    project_id="tasic_visp_scrnaseq",
)
```

```python
schema = build_arrow_schema(DataSet)
table = models_to_table([tasic_2018_ds], schema)
table = attach_linkml_metadata(table, linkml_class="DataSet")
```

```python
PATH = "../results/dataset/"
write_deltalake(PATH, table, mode="append", partition_by=["project_id"])
```

```python
inh_ps_df = pd.read_csv("../data/visp-features-and-mapping/patchseq_tx_cell_ttype_labels.csv", index_col=0)
```

```python
data_items = []
data_set_data_item_assoc = []
for i, r in inh_ps_df.iterrows():
    data_items.append(DataItem(
        id=str(i),
        name=str(i),
        project_id="visp_patchseq",
    ))
    data_set_data_item_assoc.append(DataItemDataSetAssociation(
        dataitem_id=str(i),
        dataset_id=inh_ps_ds.id,
        project_id='visp_patchseq'
    ))
```

```python
schema = build_arrow_schema(DataItem)
table = models_to_table(data_items, schema)
table = attach_linkml_metadata(table, linkml_class="DataItem")  # version auto-populated
```

```python
PATH = "../results/dataitem/"
write_deltalake(PATH, table, mode="append", partition_by=["project_id"])
```

```python
schema = build_arrow_schema(DataItemDataSetAssociation)
table = models_to_table(data_set_data_item_assoc, schema)
table = attach_linkml_metadata(table, linkml_class="DataItemDataSetAssociation")  # version auto-populated
```

```python
PATH = "../results/dataitem_dataset_association/"
write_deltalake(PATH, table, mode="append", partition_by=["project_id"])
```
