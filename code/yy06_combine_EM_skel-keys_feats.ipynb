```python
from itertools import zip_longest
def get_list_differences(list_a, list_b, buf=30):
    """
    Returns items unique to list_a and items unique to list_b.
    """
    set_a = set(list_a)
    set_b = set(list_b)
    
    only_in_a = list(set_a - set_b)
    only_in_b = list(set_b - set_a)
    
    print(f"{'ONLY IN A':<{buf}} | {'ONLY IN B':<{buf}}")
    print("-" * (2*buf))
    
    for a, b in zip_longest(only_in_a, only_in_b, fillvalue=""):
        print(f"{a:<{buf}} | {b:<{buf}}")

    print('\n')
    #return only_in_a, only_in_b
```

```python
get_list_differences(em_exc_list, visp_exc_list)
get_list_differences(em_inh_list, visp_inh_list)

get_list_differences(visp_exc_list, visp_inh_list, buf=45)
```
